# **indentation**

```python
import time  # statement 1

def count(limit):  # statement 2
    result = 0  # statement 3
    for a in range(1, limit + 1):  # statement 4
        for b in range(a + 1, limit + 1):
            for c in range(b + 1, limit + 1):
                if c * c > a * a + b * b:
                    break
                if c * c == (a * a + b * b):
                    result += 1
    return result
```
