# instance method and class method

```python
class Human:
  faults = 0  # class attribute
  # constructor method
  def __init__(self,name,age):
    self.name = name
    self.age = age
    # self.faults = faults
    print('hello from constructor method')
  #instance method
  def speak(self):
    print(f'hello, i am {self.name}')

  @classmethod
  def makefaults(cl):
    cl.faults += 1
    print(cl.faults)

  @staticmethod
  def measureTemp(temp):
    if (temp == 37):
      return 'Normal'
    else:
      return 'not normal'
```

```python
class Human:
  faults = 0  # class attribute
  # constructor method
  def __init__(self,name,age):
    self.name = name
    self.age = age
    # self.faults = faults
    print('hello from constructor method')
  #instance method
  def speak(self):
    print(f'hello, i am {self.name} and i am human')

  @classmethod
  def makefaults(cl):
    cl.faults += 1

class Employee(Human):
  def __init__(self,name,age,salary):
    # Human.__init__(self,name,age)
    super().__init__(name,age)
    self.salary = salary

  def speak(self):
    super().speak()
    print(f'hello, i am {self.name} and i am employee')


  def work(self):
    print(f'{self.name} is working')

class Teacher(Employee):
  def __init__(self,name,age,salary,subject):
    super().__init__(name,age,salary)
    self.subject = subject
  def speak(self):
    super().speak()
    print(f'i am {self.name}, and i am a teacher')
```

```python
class Human:
  faults = 0
  def __init__(self,name,age):
    self.__name = name
    self.__age = age

  def talk(self):
    print(f'hello, i am {self.__name}')

  @property
  def get_name(self):
    return self.__name
  @property
  def get_age(self):
    return self.__age

  @get_age.setter
  def set_age(self,age):
    if age < 0:
      print('invalid age')
      raise ValueError('invalid age')
    else:
      self.__age = age
```

```python
h1 = Human('lara', 10)
h1.get_name
```

```python
h1.age = -20
h1.age
```

```python
h1.set_age = -10
```
