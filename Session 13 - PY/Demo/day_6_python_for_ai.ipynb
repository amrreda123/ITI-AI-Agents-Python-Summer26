```python
data = {
"name": ["Amir", "Sara", "Leo", "Nina"],
"group": ["A", "B", "A", "B"],
"score": [90, 70, 82, 60],
}
df = pd.DataFrame(data)
print(df.groupby("group")['score'].mean())
```

```python
print(df.groupby("group")["score"].count())
```

```python
print(df.groupby("group")["score"].max())
```

```python
print(df.groupby("group")["score"].min())
```

```python
print(df.groupby("group")["score"].agg(["mean", "max", "count"]))
```
